**NUMPY RESHAPE**:
RESHAPE MEANS CHANGING THE SHAPE OF AN ARRAY


```python
# Reshape functon:reshaping means changing the shape of an array,
# the shape of an array is the number of elements in each dimension.
# Using arange() to generate numpy array x with numbers between 1 to 15
x = np.arange(1, 19)
print(x)
print(x.shape)
print(x.ndim)
```

```python
# Reshape x with  3 rows and 6 columns
n=x.reshape((3,6))
print(n)
print(n.shape)
print(n.ndim)
```

```python
n1=n.reshape((3,6,1))
print(n1)
print(n1.shape)
print(n1.ndim)
```

```python
#ravel
# convert to old dimension
v=n.ravel()
print(v.shape)
print(v)
print(v.ndim)
```
